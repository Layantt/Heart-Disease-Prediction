# Heart Disease Prediction — Feature Selection

**Project overview:** This Data Mining project analyzes patient health data to predict the presence of heart disease through classification and explore similarities among patients through clustering.

**Code sample:** The following section compares three feature selection methods to identify relevant predictors: SelectKBest, RFE, and L1 regularization. It uses the preprocessed `df_filtered` dataset from the original notebook.

[Original team project](https://github.com/JoudySaleh-IT/PredictingHeartDisease)


**Filter method — SelectKBest (ANOVA F-test)**


```python
# Separate the input features from the target column.
X = df_filtered.iloc[:, :-1]
y = df_filtered.iloc[:, -1]

# Select the two features with the highest ANOVA F-scores.
selector = SelectKBest(score_func=f_classif, k=2)
X_new = selector.fit_transform(X, y)

# Recover the names of the selected columns.
selected_features = X.columns[selector.get_support()]
print("Selected Features:", selected_features)
```

**Wrapper method — Recursive Feature Elimination (RFE)**


```python
# Separate predictors and target.
X = df_filtered.iloc[:, :-1]
y = df_filtered.iloc[:, -1]

# RFE repeatedly removes features using a logistic regression estimator.
model = LogisticRegression()
rfe = RFE(model, n_features_to_select=2)
X_new = rfe.fit_transform(X, y)

# Print the positions (not column names) of the retained features.
selected_features = [i for i, mask in enumerate(rfe.support_) if mask]
print("Selected Features:", selected_features)
```

**Embedded method — L1 Regularization (Lasso)**


```python
# Separate the input features from the target column.
X = df_filtered.iloc[:, :-1]
y = df_filtered.iloc[:, -1]

# Fit Lasso regression with L1 regularization.
model = Lasso(alpha=0.1)
model.fit(X, y)

# Retain features whose fitted coefficients are non-zero.
selected_features = X.columns[model.coef_ != 0]
print("Selected Features:", selected_features)
```

**Saved selections:** SelectKBest: `exang`, `oldpeak`; RFE: positions `[7, 9]`; Lasso: `age`, `cp`, `trestbps`, `chol`, `thalach`, `oldpeak`, `ca`.
